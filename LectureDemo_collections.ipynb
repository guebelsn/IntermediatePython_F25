### Week 8 Lecture demo - Using Python's collections module


```python
from collections import *
```

```python
# create a deque from a list of numbers
myNums = [27, 600, 42, 13, 50.5, 47, -1]
myQueue = deque(myNums)
myQueue.appendleft(13)
#myQueue.popleft()
del myQueue[2]
```

```python
myQueue
```

```python
#create a deque from a string
word = "Hello"
myChars = deque(word, maxlen = 3)
myChars.append('!')
myChars.appendleft('?')
myChars
```

```python
# create a defaultdict (with integer default)
myDictionary = defaultdict(list)
myDictionary['one'] = [1]
myDictionary['two'] = [2]
myDictionary['two'].append(3)
myDictionary['three'].append(-1)
myDictionary['three']
```

```python
newDict = defaultdict(bool)
print(newDict['nextKey'])
```

```python
salesData = { "chocolate chip": [130.50, 120, 32, 45], "macademian" : [20.5,23,5,34] }
total_revenue_dict = defaultdict(float)
print(total_revenue_dict['oatmeal raison'])

for product, revenue in salesData.items():
    total_revenue_dict[product] = sum(revenue)

total_revenue_dict
```

```python
# create a Counter from iterable
myNums = [27, 600, 42, 13, 50.5, 47, -1]*2+ [27, 42, 0, 0, 1, 2, 2,13]
myNums
myCounts = Counter(myNums)
common_list = myCounts.most_common()
common_list.reverse()
common_list[:3]
list(myCounts.elements())
myNums2 = [27,600,42,27]
myCounts2 = Counter(myNums2)
myCounts + myCounts2
```

```python
# create a Counter from initial counts
myDict = {"chocolate chip": 1, "Macademian": 4}
otherCounts = Counter(myDict)
print(otherCounts)
#Find the name of the most common
most = otherCounts.most_common(1)
most[-1][0]

```
