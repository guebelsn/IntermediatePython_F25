```python
# Creating a Table
myData_tb = Table().with_columns('Name', ['Nathalie','Alfred','Sarah'], 'Favorite num', [1, 2, 3])
myData
```

```python
farmers_markets = Table().read_table('farmers_markets.csv')
farmers_markets
```

```python
#Try where and only look at 2 columns County and city, dropping nan
farmers_markets_CA = farmers_markets.where('State', 'California').select('County','city').where('city', are.not_equal_to('nan')).where('County', are.not_equal_to('nan'))
farmers_markets_CA
```

```python
# Using group
farmers_markets_CA.group('County')
```

```python
# Clean up duplicate, different spelling of county names
farmers_markets_CA.with_column('County', farmers_markets_CA.apply(str.capitalize, 'County')).group('County')
```

#### 4. MatplotLib for visualizations
- Plot types: line, bar, scatter, histogram
- Customization: titles, labels, legends, colors
*(note that the datascience library has these directly built-in)*



```python
%matplotlib inline
import matplotlib.pyplot as plt
little_women = pd.read_csv('little_women.csv')
little_women
plt.scatter(little_women['Characters'], little_women['Periods'], c='green')
plt.xlabel('Characters')
plt.ylabel('Periods')
#plt.show() # don't need .show() in notebook if you have %matplotlib inline at the top
```

```python
farmers_markets_pd_CA = pd.read_csv('farmers_markets.csv')[farmers_markets_pd['State'] == 'California'] [['County','city']].dropna()
farmers_markets_pd_CA['County'] = farmers_markets_pd_CA['County'].str.capitalize()
farmers_markets_pd_byCounty = farmers_markets_pd_CA.groupby('County').count()
farmers_markets_pd_byCounty
```

```python
#farmers_markets_pd_byCounty.plot.barh(legend = False, figsize=(8, 10))
# or
farmers_markets_pd_byCounty.sort_values(by = 'city').plot.barh(legend = False, figsize=(8, 10))
# Add labels and title
plt.xlabel('count')
plt.ylabel('County')
plt.title('Bar Chart from number of farmers markets in each County, in CA')
# Adjust the y-axis tick width
plt.tick_params(axis='y', labelsize = 8) # Set width to 2 points
plt.show()
```

#### 5. From Data Analysis to Machine Learning - using sklearn
First steps:
- Clean, structured data feeds into ML models
- Convert Pandas DataFrames into NumPy arrays
=> sklearn works with np.array()

[sklearn documentation](https://scikit-learn.org/stable/)


```python
from sklearn.linear_model import LogisticRegression
from sklearn import datasets
# data downloaded from https://github.com/mwaskom/seaborn-data/blob/master/iris.csv
iris =  pd.read_csv('iris.csv')
#Convert DataFrame to numpy arrays
X = iris[['sepal_length','sepal_width','petal_length','petal_width']]
y = iris['species']
model = LogisticRegression(max_iter=200)
model.fit(X, y)
model
model.score(X, y)
```
